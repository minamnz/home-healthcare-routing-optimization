# Synthetic Data Generation

Generate small, illustrative problem instances for a home healthcare routing exercise. This notebook is **independent** of the solver notebooks and does not automatically transfer its outputs to them. No real patient data are used.


```python
import random
import math

def generate_multiple_small_instances(
    num_instances=5,
    distance_range=(1, 50),
    demand_range=(1, 5),
    sample_factor_range=(0.5, 1.0),
    service_time_range=(0, 2),
    earliest_time=0,
    latest_time=100,
    capacity=10,
    transport_cost=5,
    penalty=2.0,
    big_m=50,
    seed=None
):


    if seed is not None:
        random.seed(seed)

    data_list = []

    for _ in range(num_instances):

        M = 3
        N = 2
        K = 1


        Dij = []
        for i in range(M):
            row = []
            for j in range(M):
                if i == j:
                    row.append(0)
                else:
                    dist = random.uniform(*distance_range)
                    row.append(dist)
            Dij.append(row)


        Ai = [random.randint(*demand_range) for _ in range(M)]

        Bi = []
        for i in range(M):
            factor = random.uniform(*sample_factor_range)
            Bi.append(Ai[i] * factor)


        Ei = [earliest_time]*M
        Li = [latest_time]*M
        Wi = [random.randint(*service_time_range) for _ in range(M)]


        CAPk_data = [capacity]
        TCk_data  = [transport_cost]
        PEN       = penalty
        BIG       = big_m


        data = {
            'M': M,
            'N': N,
            'K': K,
            'Dij': Dij,
            'Ai': Ai,
            'Bi': Bi,
            'Ei': Ei,
            'Li': Li,
            'Wi': Wi,
            'CAPk': CAPk_data,
            'TCk': TCk_data,
            'PEN': PEN,
            'BIG': BIG
        }

        data_list.append(data)

    return data_list



if __name__ == "__main__":

    instances = generate_multiple_small_instances(
        num_instances=3,
        distance_range=(5, 20),
        demand_range=(2, 5),
        sample_factor_range=(0.3,1),
        service_time_range=(0,2),
        earliest_time=0,
        latest_time=100,
        capacity=10,
        transport_cost=5,
        penalty=1.5,
        big_m=50,
        seed=42
    )


    for idx, data in enumerate(instances, start=1):
        print(f"\n=== Instance #{idx} ===")
        for key, val in data.items():
            print(f"{key} => {val}")
```
